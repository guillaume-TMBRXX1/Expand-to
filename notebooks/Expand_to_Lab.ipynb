## Carte

```python
import matplotlib.pyplot as plt
from matplotlib.colors import to_rgb
from expandto import generate_map, load_data

data = load_data()
SEED = 4242
grid = generate_map(SEED, data['rules']['map']['width'], data['rules']['map']['height'])
img = [[to_rgb(data['terrain'][t['type']]['color']) for t in row] for row in grid]
plt.figure(figsize=(10, 7)); plt.imshow(img); plt.title(f'Graine {SEED}'); plt.axis('off'); plt.show()

from collections import Counter
Counter(t['type'] for row in grid for t in row)
```
